```python
# BAI 1: TAO DATAFRAME

import pandas as pd

data = {
    "Ho_ten": [
        "Thinh", "Binh", "Dung", "Khoi", "Tin",
        "Lan", "Minh", "Nam", "Phuc", "Trang"
    ],
    "Chuyen_can": [8, 9, 7, 10, 8, 6, 9, 7, 8, 10],
    "Giua_ky":    [7, 8, 6, 9, 8, 7, 9, 8, 7, 9],
    "Cuoi_ky":    [8, 9, 7, 10, 8, 6, 9, 7, 8, 10]
}

df = pd.DataFrame(data)

display(df)
```

```python
# BAI 2: TINH DIEM TONG KET

df["Tong_ket"] = (
    0.2 * df["Chuyen_can"]
    + 0.3 * df["Giua_ky"]
    + 0.5 * df["Cuoi_ky"]
)

display(df.round(2))
```

```python
# BAI 3: XEP LOAI HOC LUC

def xep_loai(diem):
    if diem >= 8.5:
        return "Gioi"
    elif diem >= 7.0:
        return "Kha"
    elif diem >= 5.0:
        return "Trung binh"
    else:
        return "Yeu"

df["Xep_loai"] = df["Tong_ket"].apply(xep_loai)

display(df.round(2))
```

```python
# BAI 4: THONG KE KET QUA

# Diem trung binh cua lop
print("Diem trung binh cua lop:",
      round(df["Tong_ket"].mean(), 2))

# Diem cao nhat
print("Diem cao nhat:",
      df["Tong_ket"].max())

# Diem thap nhat
print("Diem thap nhat:",
      df["Tong_ket"].min())

# So sinh vien dat (diem >= 5)
print("So sinh vien dat:",
      (df["Tong_ket"] >= 5).sum())

# Danh sach sinh vien diem cao nhat
print("Sinh vien co diem cao nhat:")
display(df[df["Tong_ket"] == df["Tong_ket"].max()])

# Danh sach sinh vien diem thap nhat
print("Sinh vien co diem thap nhat:")
display(df[df["Tong_ket"] == df["Tong_ket"].min()])
```

```python
# BAI 5: VE BIEU DO DIEM TONG KET

import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.bar(df["Ho_ten"], df["Tong_ket"])

plt.title("DIEM TONG KET CUA 10 SINH VIEN")
plt.xlabel("Ho ten sinh vien")
plt.ylabel("Diem tong ket")

plt.ylim(0, 10)
plt.grid(axis="y", alpha=0.3)

plt.show()
```
